## 2. Loading Pretrained Transformers & Tokenizers

### Conceptual Overview: Subword Tokenization & Transformer Heads
* **Subword Tokenizer (`AutoTokenizer`)**: Uses WordPiece/BPE algorithms to tokenize text into vocabulary subwords and adds special tokens (`[CLS]`, `[SEP]` for BERT).
* **Transformer Classification Head (`AutoModelForSequenceClassification`)**: Attaches a linear classification layer on top of the pooled `[CLS]` token representation.


```python
# Inspect and load Hugging Face Transformers
try:
    from transformers import AutoTokenizer, AutoModelForSequenceClassification, pipeline
    print('[Success] Hugging Face Transformers is installed and ready!')
    
    # Initialize zero-shot sentiment analysis pipeline
    classifier = pipeline('sentiment-analysis', model='distilbert-base-uncased-finetuned-sst-2-english')
    
    sample_texts = [
        'Deep learning and transformers are revolutionizing artificial intelligence!',
        'The execution was slow and buggy.'
    ]
    
    # Perform inference
    results = classifier(sample_texts)
    for text, res in zip(sample_texts, results):
        print(f'Text: "{text}" -> Label: {res["label"]} (Confidence: {res["score"]:.4f})')
        
except ImportError:
    print('Transformers library not installed. Install via: pip install transformers')
```
